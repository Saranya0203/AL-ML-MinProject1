```python
import requests
import pandas as pd
```

```python
metadata_url = "https://data.cityofnewyork.us/api/views/erm2-nwe9.json"
response=requests.get(metadata_url)
response.raise_for_status()
metadata=response.json()
```

```python
columns_df=pd.DataFrame([
    {
      "Column Name":col.get("name"),
      "API Field Name":col.get("fieldName"),
      "Data Typ":col.get("dataTypeName"),
      "Description":col.get("description"),
    }
for col in metadata["columns"]
])
```

```python
columns_df
```

```python
available_columns = columns_df["API Field Name"].tolist()
```

```python
available_columns
```

```python
selected_columns = [
    "unique_key",
    "created_date",
    "closed_date",
    "agency",
    "agency_name",
    "complaint_type",
    "descriptor",
    "descriptor_2",
    "location_type",
    "incident_zip",
    "incident_address",
    "city",
    "status",
    "due_date",
    "resolution_description",
    "resolution_action_updated_date",
    "community_board",
    "council_district",
    "police_precinct",
    "borough",
    "latitude",
    "longitude",
    "open_data_channel_type",
    "park_facility_name",
    "facility_type"
]

```

```python
data_url = "https://data.cityofnewyork.us/resource/erm2-nwe9.json"

row_limit = 100000
batch_size = 5000  # SODA supports up to 5000 per request
all_records = []

for offset in range(0, row_limit, batch_size):
    parameters = {
        "$select": ",".join(selected_columns),
        "$limit": batch_size,
        "$offset": offset,
        "$order": "created_date DESC"
    }
    response = requests.get(data_url, params=parameters, timeout=30)
    response.raise_for_status()
    batch_record = response.json()
    all_records.extend(batch_record)

    print(f"Records collected so far: {len(all_records)}")

    if len(batch_record) < batch_size:
        print("No more records returned from API.")
        break

# Convert to DataFrame
df_311 = pd.DataFrame(all_records)
display(df_311.head())
```
